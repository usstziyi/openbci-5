在 **BrainFlow** 中，`BoardShim` 是核心类， **静态方法 / 类方法**，主要可以分为几个类别：

---

### 1️⃣ **板卡描述 / 元信息查询**

| API                                     | 说明                            | 返回值类型 |
| --------------------------------------- | ----------------------------- | ----- |
| BoardShim.get_board_descr(board_id)   | 获取板卡的完整描述字典，包括 EEG、加速度通道、采样率等 | dict  |
| BoardShim.get_eeg_names(board_id)     | 获取 EEG 通道名列表（固定列表）            | list  |
| BoardShim.get_accel_names(board_id)   | 获取加速度通道名                      | list  |
| BoardShim.get_analog_names(board_id)  | 获取模拟输入通道名                     | list  |
| BoardShim.get_other_names(board_id)   | 获取其他辅助通道名                     | list  |
| BoardShim.get_sampling_rate(board_id) | 获取板卡采样率                       | int   |
| BoardShim.get_num_rows(board_id)      | 获取每次采样的数据行数（总通道数）             | int   |

---

### 2️⃣ **数据操作 / 采集相关**

> 这些通常是 **实例方法**，不是类方法，需要先 `board = BoardShim(...)`

| API                                  | 说明           |
| ------------------------------------ | ------------ |
| board.prepare_session()            | 初始化板卡，建立连接   |
| board.start_stream()               | 开始数据采集       |
| board.get_board_data(n_samples=-1) | 获取缓存区 EEG 数据 |
| board.stop_stream()                | 停止采集         |
| board.release_session()            | 释放资源，关闭连接    |

---

### 3️⃣ **实用工具类方法**

| API                                   | 说明                               |
| ------------------------------------- | -------------------------------- |
| BoardShim.enable_dev_board_logger() | 打开 BrainFlow 内部日志，用于调试           |
| BoardShim.disable_board_logger()    | 关闭日志                             |
| BoardShim.set_log_level(level)      | 设置日志等级（DEBUG, INFO, WARN, ERROR） |
| BoardShim.get_board_id_list()       | 获取 BrainFlow 已知所有板卡 ID 列表        |
| BoardShim.get_board_name(board_id)  | 获取板卡名称字符串                        |
| BoardShim.get_version()             | 获取 BrainFlow SDK 版本              |

---

💡 **总结**：

* **类方法 / 静态方法**：用于 **查询板卡信息或 BrainFlow SDK 状态**，无需实例化 `BoardShim`
* **实例方法**：用于 **实际操作板卡和采集数据**，需要先用 `BoardShim(board_id, params)` 创建对象
