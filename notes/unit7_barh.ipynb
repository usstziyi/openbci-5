```python
import matplotlib.pyplot as plt

metrics = ['Restfulness', 'Attention']
values = [0.8, 0.65]
y_pos = [0, 2]  # 自定义条形中心位置

fig, ax = plt.subplots()
bars = ax.barh(y_pos, values, color=['green', 'purple'], alpha=0.7)
ax.set_yticks(y_pos)
ax.set_yticklabels(metrics)
ax.set_xlim(0, 1)
ax.set_xlabel('Score')
plt.show()
```
