```python
import matplotlib.pyplot as plt
import matplotlib

# 设置中文字体
matplotlib.rcParams['font.sans-serif'] = [
    'Microsoft YaHei',
    'SimHei',
    'DejaVu Sans'
]
matplotlib.rcParams['axes.unicode_minus'] = False # 解决负号显示问题，确保中文显示正常

# 创建测试数据
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# 创建图表
plt.figure(figsize=(8, 6))
plt.plot(x, y, marker='o', linewidth=2, markersize=8)

# 设置标题和标签
plt.title('中文显示测试', fontsize=16)
plt.xlabel('横轴（X轴）', fontsize=12)
plt.ylabel('纵轴（Y轴）', fontsize=12)

# 添加网格
plt.grid(True, linestyle='--', alpha=0.7)

# 添加图例
plt.legend(['测试数据'], loc='upper left')

# 显示图表
plt.tight_layout()
plt.show()

print("如果图表中正确显示中文，说明matplotlib中文支持正常！")
```
