在 **BrainFlow** 的 `BoardShim` 中，`get_board_data()` 的行为是**会清空内部缓冲区的已取数据**。具体说明如下：

1. **`get_board_data()`**

   * 每次调用时，它会**从内部缓冲区读取所有可用数据**（或者根据参数限制的数量）。
   * 读取之后，这些数据在缓冲区中会被删除，也就是说缓冲区的内容会被清空或滚动，取出的数据不会保留在缓冲区中。
   * 因此，如果你调用 `get_board_data()` 多次，每次返回的数据是**自上次调用之后新积累的部分**。

2. **`get_current_board_data(n_samples)`**

   * 与 `get_board_data()` 不同，它**不会删除缓冲区的数据**。
   * 调用时只读取指定数量 `n_samples` 的数据（如果缓冲区不够就返回当前所有），数据仍然保留在内部缓冲区中供下一次调用使用。
   * 适合做实时绘图或循环处理，而不丢失未处理的数据。

3. **应用场景建议**

   * 如果你希望“拉取后不保留”，用 `get_board_data()`。
   * 如果做实时可视化或处理循环，需要保持数据在缓冲区中，多次访问同一数据段，用 `get_current_board_data()` 更安全。

总结：**`get_board_data()` 每次取数据后会清空对应缓冲区，而 `get_current_board_data()` 则不会清空缓冲区。**
