### EEG to Deep Learning
|需求|推荐库|定位|
|--|--|--|
|EEG 预处理、滤波、分段、ICA	|MNE-Python	|EEG/MEG 标准分析库|
|PyTorch 训练 EEG 深度学习	|Braindecode	|最接近“EEG + PyTorch”的深度学习库|
|EEG 深度学习模型集合	|TorchEEG	|PyTorch EEG 模型/数据集/训练流程|
|公共 EEG 数据集和 benchmark	|MOABB	|BCI 数据集与标准评测|
|实时采集硬件数据	|BrainFlow	|板卡采集、基础 DSP|
```Text
最常见的工作流是：

BrainFlow 采集
   ↓
MNE-Python 预处理
   ↓
Braindecode / TorchEEG 建模
   ↓
PyTorch 训练

重点推荐：

1. MNE-Python
用来做 EEG 数据清洗、滤波、重参考、epoch 切片、ICA、事件标记等。它不是深度学习库，但几乎是 Python EEG 处理的基础设施。 ￼

2. Braindecode
这是最像你要找的库：官方定位就是 PyTorch toolbox for decoding raw EEG/ECoG/MEG with deep learning。它内置 EEGNet、Deep4Net、ShallowFBCSPNet 等模型，也能直接接 MNE 数据结构。 ￼

3. TorchEEG
更偏“深度学习工程化”，提供数据集封装、transform、trainer，以及 CNN、GCN、Transformer 等 EEG 模型。适合做情绪识别、睡眠分期、脑机接口分类等任务。 ￼

4. MOABB
如果你要找公开 EEG 数据集，尤其是运动想象 BCI，MOABB 很有用。它主要用于可复现实验和 benchmark，可以和 MNE、Braindecode、TorchEEG 配合。 ￼

结论：

如果你用 PyTorch：BrainFlow + MNE + Braindecode / TorchEEG
如果你用 TensorFlow：BrainFlow + MNE + 自己写 tf.keras 模型

TensorFlow 生态里没有像 Braindecode 这样特别主流的 EEG 专用库，所以 EEG 深度学习目前更推荐 PyTorch 路线。
```
