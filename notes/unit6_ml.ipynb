```text
BrainFlowMetrics 和 BrainFlowClassifiers 是 brainflow.ml_model 里的两个枚举，主要用来配置 MLModel。

1. BrainFlowMetrics

表示你想计算哪一种 EEG 高级指标。

常见有：

BrainFlowMetrics.CONCENTRATION
BrainFlowMetrics.RELAXATION

含义大致是：

枚举	含义
CONCENTRATION	专注度指标
RELAXATION	放松度指标

BrainFlow 文档说明，MLModel 用来计算 derivative metrics，也就是从原始信号/频带特征进一步推导出的高级指标。早期 BrainFlow ML 模块支持的就是 concentration 和 relaxation。 ￼

⸻

2. BrainFlowClassifiers

表示用哪一种模型/分类器来计算这个指标。

常见有：

BrainFlowClassifiers.DEFAULT_CLASSIFIER

有些版本也可能支持 ONNX 相关模型配置。BrainFlow 后续加入过 ONNX 模型推理能力，可以加载部分 ONNX 模型进行跨语言推理。 ￼

一般你日常用内置模型时，用：

BrainFlowClassifiers.DEFAULT_CLASSIFIER.value

就够了。

⸻

典型用法

from brainflow.ml_model import MLModel, BrainFlowMetrics, BrainFlowClassifiers, BrainFlowModelParams
params = BrainFlowModelParams(
    BrainFlowMetrics.CONCENTRATION.value,
    BrainFlowClassifiers.DEFAULT_CLASSIFIER.value
)
model = MLModel(params)
model.prepare()
score = model.predict(feature_vector)
model.release()

这里：

BrainFlowMetrics.CONCENTRATION.value

表示：我要算“专注度”。

BrainFlowClassifiers.DEFAULT_CLASSIFIER.value

表示：使用 BrainFlow 默认模型。

⸻

它们之间的关系

可以理解为：

BrainFlowMetrics      决定“算什么”
BrainFlowClassifiers  决定“用什么模型算”

例如：

CONCENTRATION + DEFAULT_CLASSIFIER
= 用默认模型计算专注度
RELAXATION + DEFAULT_CLASSIFIER
= 用默认模型计算放松度

一句话总结：

BrainFlowMetrics 是目标指标枚举；
BrainFlowClassifiers 是模型/分类器枚举；
两者一起组成 BrainFlowModelParams，交给 MLModel 使用。
```
